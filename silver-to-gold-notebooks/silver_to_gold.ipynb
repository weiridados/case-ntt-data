QUESTÕES

What country(s) use more kind of vaccines

```python
%sql
CREATE OR REPLACE TABLE dtf_case_ntt_data.gold.question_1_country_vaccine_coverage
USING DELTA
AS

SELECT

    iso_code,

    location AS country,

    SIZE(SPLIT(vaccines, ',')) AS total_vaccine_types

FROM vw_locations

ORDER BY total_vaccine_types DESC

LIMIT 10;
```

QUESTION 2
Top 10 countries with the highest number of vaccinations

```python
%sql
CREATE OR REPLACE TABLE dtf_case_ntt_data.gold.question_2_top_vaccinations
USING DELTA
AS

SELECT

    iso_code,

    country,

    YEAR(vaccination_date) AS vaccination_year,

    MONTH(vaccination_date) AS vaccination_month,

    SUM(total_vaccinations) AS total_vaccinations

FROM vw_vaccinations

WHERE LOWER(country) NOT IN (

    'world',
    'asia',
    'africa',
    'europe',
    'north america',
    'south america',
    'oceania',
    'european union'

)

AND iso_code <> 'OWID_UMC'

GROUP BY

    iso_code,
    country,
    YEAR(vaccination_date),
    MONTH(vaccination_date)

ORDER BY total_vaccinations DESC

LIMIT 10;
```

QUESTION 3

```python
%sql
CREATE OR REPLACE TABLE dtf_case_ntt_data.gold.question_3_country_vaccines
USING DELTA
AS

SELECT

    v.iso_code,

    v.country,

    YEAR(v.vaccination_date) AS vaccination_year,

    l.vaccines,

    SIZE(SPLIT(l.vaccines, ',')) AS total_vaccine_types,

    SUM(v.total_vaccinations) AS total_vaccinations

FROM vw_vaccinations v

INNER JOIN vw_locations l
    ON v.iso_code = l.iso_code

WHERE LOWER(v.country) NOT IN (

    'world',
    'asia',
    'africa',
    'europe',
    'north america',
    'south america',
    'oceania',
    'european union',
    'upper middle income',
    'lower middle income'

)

AND v.iso_code NOT IN (

    'OWID_UMC',
    'OWID_LMC',
    'OWID_HIC',
    'OWID_LIC'

)

GROUP BY

    v.iso_code,
    v.country,
    YEAR(v.vaccination_date),
    l.vaccines

ORDER BY

    total_vaccine_types DESC,
    total_vaccinations DESC

LIMIT 10;
```
